```python
import numpy as np
import pandas as pd

np.random.seed(42)

# -----------------------------------
# 1. Metadata for realistic behavior
# -----------------------------------
channels = {
    "Programmatic": {
        "sources": ["StackAdapt", "Amazon Ad Server"],
        "ctr_range": (0.006, 0.013),
        "cpc_range": (1.10, 1.30),
        "conv_range": (0.010, 0.018)
    },
    "Paid Search": {
        "sources": ["Google", "Bing"],
        "ctr_range": (0.035, 0.047),
        "cpc_range": (0.80, 1.05),
        "conv_range": (0.025, 0.035)
    },
    "Paid Social": {
        "sources": ["Facebook", "LinkedIn Ads"],
        "ctr_range": (0.020, 0.030),
        "cpc_range": (0.65, 4.50),
        "conv_range": (0.015, 0.025)
    },
    "Organic": {
        "sources": ["Google", "Bing"],
        "ctr_range": (0.00, 0.00),
        "cpc_range": (0.00, 0.00),
        "conv_range": (0.00, 0.00)
    }
}

campaigns = [
    "Winter Promo 2024", "Brand Awareness Q1", "Search Boost", "Holiday Retargeting",
    "Lead Gen Q1", "Spring Launch", "Remarketing Q1", "Product Launch X",
    "Awareness Push", "Retention Q1"
]

# -----------------------------------
# 2. Generate 10,000 rows
# -----------------------------------
num_rows = 10_000
rows = []

for _ in range(num_rows):

    date = pd.Timestamp("2024-01-01") + pd.Timedelta(days=np.random.randint(0, 365))
    campaign = np.random.choice(campaigns)

    channel = np.random.choice(list(channels.keys()))
    source = np.random.choice(channels[channel]["sources"])

    ctr = np.random.uniform(*channels[channel]["ctr_range"])
    cpc = np.random.uniform(*channels[channel]["cpc_range"])

    impressions = np.random.randint(40_000, 250_000)
    clicks = int(impressions * ctr)
    spend = round(clicks * cpc, 2)

    conv_rate = np.random.uniform(*channels[channel]["conv_range"])
    conversions = int(clicks * conv_rate)

    cpa = round(spend / conversions, 2) if conversions > 0 else 0
    revenue = round(conversions * np.random.uniform(60, 120), 2)
    roas = round(revenue / spend, 2) if spend > 0 else 0

    rows.append([
        date.date(), campaign, channel, source,
        impressions, clicks, round(ctr * 100, 2),
        round(cpc, 2), spend, conversions, cpa, revenue, roas
    ])

# -----------------------------------
# 3. Create DataFrame
# -----------------------------------
df = pd.DataFrame(rows, columns=[
    "date", "campaign", "channel", "source",
    "impressions", "clicks", "ctr_percent",
    "cpc", "spend", "conversions", "cpa", "revenue", "roas"
])

print(df.head())
df.to_csv("C:\\Users\\Cristian\\OneDrive\\Documents\\Cristian Avila\\Tripleten\\Career Prep Course\\marketing_data.csv", index=False)
```
